```python
adv = list('ABCDEF')
addr = list('NMOP')

book = [ str(x) + ' ' +
  random.choice(adv) + ' ' + 
  random.choice(addr) + '. ' + 
  str(random.randint(11111, 99999))
  for x in range(20) ]
book = '\n'.join(book)
print(book)
```

```python
pattern = re.compile('\n+')
sname = pattern.split(book)
sname
```

```python
delFirstNum = r'\d+ '
print(re.sub(delFirstNum, '', book))
```
