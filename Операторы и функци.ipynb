### Наследование

```python
class Memes :
    
    def __init__(self, age, stale_joke = False) :
        self.age = age
        self.sj = stale_joke
        self.fresh = 10
        
    def check_sj(self) :
        self.sj = self.age > self.fresh
        
    def rise_age(self, age = 1) :
        self.age += age
        self.check_sj()
        
    def is_stale_joke(self,) :
        return self.sj
```

```python
wolfs = Memes(4, False)
wolfs.is_stale_joke()
```

```python
wolfs.rise_age(7)
wolfs.is_stale_joke()
```

```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
```

```python
class FrogMemes (Memes) :
    def __init__(self, age) :
        Memes.__init__(self, age, stale_joke = False)

    def check_sj(self) : 
        self.sj = False
        
    def doKva(self = None) :
        #print("It's Wednesday, my dudes")
        img = mpimg.imread('frooog.jpg')
        imgplot = plt.imshow(img)
        plt.xticks([])
        plt.yticks([])
```

```python
frog = FrogMemes(11)
print('Is frog a stale joke? Absolutly', 'YES' if frog.is_stale_joke() else 'NO')  
frog.doKva()
```

```python
frog.rise_age(10)
print('Ten years ago ... \nIs frog a stale joke? Absolutly', 'YES' if frog.is_stale_joke() else 'NO')  
```
