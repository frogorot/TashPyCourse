# Профилирование

```python
A = np.array([[1,2,3],[4,5,6]])
rng = np.random.default_rng()
B=rng.random((1000,1000))
print(B.shape)
```

```python
def fli(arr) :
    C = arr.copy()
    C = np.transpose(C)
    print(C.shape)
    return C
    
def sli(arr) :
    C = np.empty((arr.shape[1], arr.shape[0]))
    print(C.shape)
    for i in range(arr.shape[0]) :
        for j in range(arr.shape[1]) :
            C[j,i]=arr[i,j]
    return C
```

```python
%time fli(B)
print()
%time sli(B);
```

# Посчитаем что-то ~~полезное~~ на numpy

```python
b=np.linspace(0,20, num=11, dtype=np.float64)
b = np.concatenate((b, b[::-1]))
print(b)
print(b.sum())

def filter(arr) :
    c = arr.copy()
    
    c[1:-1] = arr[0:-2] /4 + arr[1:-1] /2 + arr[2:] /4
    
    return c
```

```python
def use (item, count) :
    for i in range(count) :
        item = filter(item)
    return item

jobs = [ b ] 
plt.plot(jobs[0])

for x in range(1, 20) :
    jobs.append(use(jobs[x-1], x*20))
    plt.plot(jobs[x])
```

# Матричная свёртка

```python
pic = np.random.randint(0,256, (30,30))
plt.imshow(pic, cmap = plt.cm.gray)
plt.show()
```

```python

def filter(picture) :
    
    pic = picture.copy()
    
    pic[1:-1, 1:-1] = (2*pic[0:-2, 0:-2] - pic[0:-2, 2:] +
                        2*pic[1:-1, 0:-2]  - 2*pic[1:-1, 2:] +
                        2*pic[2:, 0:-2]  - pic[2:, 2:])
    return pic
    
plt.imshow(filter(pic), cmap = plt.cm.gray)
#core = np.arange([ [1,0,-1],
#                 [2, 0, -2],
#                 [1,0,-1]])
```
